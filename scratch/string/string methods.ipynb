```python
''' String methods are built in functions - they are used to change/modify the string '''
```

```python
#.upper() - Upper case
word = "Cat"
print(word.upper())
```

```python
#.lower() - Lower case
print(word.lower())
```

```python
#.strip() - Remove extra space from left or right
sentence = "        Northern Lights are amazing  "
print(sentence.strip())
print(sentence.lstrip())
print(sentence.rstrip())

#remove particular characters
x="@@@@Lion"
print(x.strip("@"))
```

```python
#.capitalize() - Capital the first letter
color = "blue"
print(color.capitalize())
```

```python
#.title() - capital's each word in a string
movie = "Harry Potter"
print(movie.title())
```

```python
#.replace()
animals = "Cats are cute"
print(animals.replace("Cats","Dogs"))
```
